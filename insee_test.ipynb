```python
import pandas as pd
import json
import requests
import geopandas as gpd

communes_littorales = gpd.read_file("../data/communes_littorales/communes_littorales.shp")
communes_littorales['INSEE_COM'] = communes_littorales['INSEE_COM'].astype(str).str.zfill(5)

df_total = pd.DataFrame()
for dep in communes_littorales['INSEE_DEP'].unique():
    print(dep, len(communes_littorales[communes_littorales['INSEE_DEP'] == dep]))
    comm_list = communes_littorales[communes_littorales['INSEE_DEP'] == dep]['INSEE_COM'].tolist()


    # Construction de l'URL avec toutes les communes de comm_list
    geo_params = "&".join([f"GEO=COM-{code}" for code in comm_list])
    api_url = f"https://api.insee.fr/melodi/V2/data/DS_FILOSOFI_CC?{geo_params}&FILOSOFI_MEASURE=IR_D9_D1_SL"


    get_data = requests.get(api_url, verify= False)
    data_from_net = get_data.content
    try:
        data = json.loads(data_from_net)
    except:
        continue
    # Extraction des informations du jeu de données
    title = data['title']['fr']
    identifier = data['identifier']

    #Extraction des observations du jeu de données filtré, sur lesquelles on va boucler
    observations = data['observations']
    extracted_data = []

    #Boucle de lecture des observations dans le json 
    for obs in observations:
        dimensions = obs['dimensions']
        
        # Suivant les jeux de données attributes est présent ou non
        if 'attributes' in obs:
            attributes = obs['attributes']
        else:
            attributes = None

        # Suivant les jeux de données value peut être absent
        if 'value' in obs['measures']['OBS_VALUE_NIVEAU']:
            measures = obs['measures']['OBS_VALUE_NIVEAU']['value']
        else:
            mesures = None
        
        # on rassemble tout dans un objet
        if 'attributes' in obs:
            combined_data = {**dimensions,**attributes, 'OBS_VALUE_NIVEAU': measures}
        else:
            combined_data = {**dimensions, 'OBS_VALUE_NIVEAU': measures}
        
        extracted_data.append(combined_data)

    #Création d'un dataframe python
    df = pd.DataFrame(extracted_data)
    df_total = pd.concat([df_total, df], ignore_index=True)

    # print(f'Jeu de données : {identifier} \nTitre : {title} ')
    print('Loaded', len(df), 'rows for dep', dep)
```

```python
from math import ceil

dep = 17
communes_dep = communes_littorales[communes_littorales['INSEE_DEP'] == str(dep).zfill(2)]['INSEE_COM'].tolist()
missing_communes = [com for com in communes_dep if com not in df_total['GEO'].str.split('-').str[-1].tolist()]


batch_size = 8
num_batches = ceil(len(missing_communes) / batch_size)

for i in range(num_batches):
    batch = missing_communes[i*batch_size:(i+1)*batch_size]
    if not batch:
        continue
    geo_params = "&".join([f"GEO=COM-{com}" for com in batch])
    url_api = f"https://api.insee.fr/melodi/V2/data/DS_FILOSOFI_CC?{geo_params}&FILOSOFI_MEASURE=IR_D9_D1_SL"
    print(f"Batch {i+1}/{num_batches}: {batch}")
    get_data = requests.get(url_api, verify=False)
    data_from_net = get_data.content
    try:
        data = json.loads(data_from_net)
    except:
        print("Failed to load data for batch:", batch)
        continue
    if 'observations' not in data:
        print("No observations for batch:", batch)
        continue
    observations = data['observations']
    extracted_data = []
    for obs in observations:
        dimensions = obs['dimensions']
        attributes = obs.get('attributes', None)
        if 'value' in obs['measures']['OBS_VALUE_NIVEAU']:
            measures = obs['measures']['OBS_VALUE_NIVEAU']['value']
        else:
            measures = None
        if attributes:
            combined_data = {**dimensions, **attributes, 'OBS_VALUE_NIVEAU': measures}
        else:
            combined_data = {**dimensions, 'OBS_VALUE_NIVEAU': measures}
        extracted_data.append(combined_data)
    if extracted_data:
        df = pd.DataFrame(extracted_data)
        df_total = pd.concat([df_total, df], ignore_index=True)
```
